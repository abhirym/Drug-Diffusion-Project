## I. Gefitinib Molecular Structure & Transport Implications

There are many noteworthy functional groups and substituents to mention relating to the transport through a medium (tumor tissue, ECM).

### 1. Morpholine Ring

The morpholine ring is a weak base that serves as a soluble tail for the whole molecule, with the ability to get protonated on the tertiary amine group. We can calculate the ratio of morpholine groups that are neutral to protonated. Using the Henderson Hasselbalch equation,

$$Morpholine ⇋ Morpholine-H⁺$$

$$pH = pKₐ + log(\frac{[Morpholine]}{[Morpholine-H⁺]})$$

The typical morpholine ring has a pKa = ~8.5, and the physiological pH = 7.4. Plugging these values in and solving for the ratio,

$$7.4 = 8.5 + log(\frac{[Morpholine]}{[Morpholine-H⁺]})$$
$$-1.1=log(\frac{[Morpholine]}{[Morpholine-H⁺]})$$
$$0.0794= \frac{[Morpholine]}{[Morpholine-H⁺]}$$

This roughly translates to a ratio of 1 Morpholine : 12.6 Morpholine-H⁺.

The extremely low ratio of neutral : protonated morpholine impacts the permeability of a large proportion of gefitinib because the positive charge doesn’t allow for passive membrane diffusion, so many molecules do not reach the interior of tumor cells.

### 2. Quinazoline Scaffold

The quinazoline scaffold provides a hydrophobic center for the entire molecule, improving the transport through the cellular membrane. Although, if the morpholine group is deprotonated, the central non-nitrous benzene ring can weakly bond in several ways with hydrophobic proteins, creating an overall strong interaction that can slow transport. This state also increases the lipophilicity of the entire molecule (LogP = 3.1 - 4.1), meaning it will easily interact with membrane lipids.

### 3. Halogen Substituents

The fluoro and chloro substituents on the other phenyl ring add lipophilicity to the entire molecule. Water cannot interact with these halogens using hydrogen bonding to stay in a high-entropy state, thus it forms a stable structure between other water molecules, reducing overall entropy which is unfavorable. 

Because of this unfavorability, the halogens prefer hydrophobic environments like the cell membrane, so membrane diffusion becomes passive. Yet, the increased lipophilicity has downsides when gefitinib is moving through the bloodstream.

## II. Extracellular Factors/Determinants

On the other hand, diffusion in the extracellular matrix is improved with the protonated morpholine being able to form ion-dipole interactions with the oxygens in the water lying in between cells. This makes the molecule soluble in ECM fluid

Two important proteins in this section are HSA (Human Serum Albumin) and AAG (Alpha-1-acid-Glycoprotein), both drug transport proteins in the bloodstream. Normally, these proteins wouldn’t escape into the surrounding cells, but can leak through to the extracellular matrix and accumulate if present in a tumor. Fortunately, HSA is a nutrient for tumors and cancers, so most of it gets degraded into amino acids for other purposes, but AAG is strongly attracted to the halogen substituents and can bind to free gefitinib:

$$ Gefitinib + AAG ⇋ Gefitinib-AAG $$

One reason this attraction might be possible is due to halogen bonding. Halogen bonding comes from a polar bond with a halogen (usually chlorine, bromine or iodine). Even though the halogen is partially negative, there is a portion on the opposite side of the bond that has a partial positive charge called a **sigma hole**.

This hole can attract highly electronegative atoms found on amino acid side chains in large proteins (like AAG), meaning the chlorine substituent could hypothetically attract AAG side chains that could form a halogen bond interaction.

## III. Intracellular Factors/Determinants

The halogens increases the lifespan for gefitinib by resisting quick enzyme metabolism, and allows for sustained exposure in the tumor. **Cytochrome P450** enzymes often catalyze phenyl hydroxylation reactions, turning C-H into C-OH for exertion. This reaction can’t be easily catalyzed when a carbon-halogen bond is present, so the gefitinib molecule stays in the cell longer.

Gefitinib can also be trapped inside acidic lysosomes (pH = ~6.5 - 6.8) that induce a protonation, reducing the concentration of free gefitinib to reach EGFR molecules and inhibit the signal transduction.

When gefitinib inhibits EGFR (causing a conformational change in the kinase receptor), the **MAPK/ERK** pathway is suppressed. Lower ERK signaling means less **Cyclin D** expression, and less **CDK4/6** complex activity. Progression of the cell-cycle and proliferation rates slow down.

## IV. Connection to Spatial Gradient Modeling

**1. Molecular Properties → Diffusion Coefficient ($D$)**

The physicochemical properties of gefitinib discussed in Section I influence the diffusivity of the molecule, and change the diffusion coefficient from Fick’s Second Law.

* Lipophilicity is increased with the halogenated phenyl ring and quinazoline scaffold
* Higher lipophilicity creates easier partitioning of the cell membrane
* Yet, high hydrophobicity causes strong, non-specific attraction to proteins and ECM components
* **Model relevance :** 
    * $D_{eff}$ (effective diffusion coefficient) < $D_{H₂O}$, transport through biological tissue is slower than transport in solution

**2. ECM Heterogeneity → Spatially Variable Diffusion**

* Not all tumor tissue is uniform
* Varying ECM densities, proximity to blood vessels
* **Model relevance :**
    * Instead of a constant diffusion coefficient D, using a function D(x) that depends on tumor depth (mm)

**3. Protein Binding → Reduced Free Drug Concentration ($C_{free}$)**

* Not all gefitinib is free to bind to EGFR sites
* $C_{total} = C_{free} + C_{bound}$
* Proteins that bind with gefitinib decrease $C_{free}$
* **Model relevance :**
    * The gefitinib concentration in the PDE solutions is $C_{free}$, not $C_{total}$

**4. Tumor pH → Concentration Availability**

* The morpholine protonation creates two charge states for gefitinib: 
    * The neutral form can cross cell membranes easily
    * The charged form (+) is soluble in extracellular fluid, but has poor membrane diffusion
* Cellular gefitinib concentration (neutral) is dependent on the tumor pH and extracellular concentration (charged)
* **Model relevance :**
    * The above condition can be modeled using:

$$C_{cellular} = ƒ(pH) * C_{extracellular}$$

**5. Intracellular Transport → EGFR Inhibition Model**

* Uses a Hill inhibition equation for EGFR activity following gefitinib exposure
* $C(x,t)$ (spatial gefitinib concentration) → $A_{EGFR}(x, t)$ (EGFR spatial activity)
* **Model relevance :**
    * Accurately connecting gefitinib concentration to EGFR activity is the first step to mapping Gefitinib → Proliferation
