## Problem Statement

Targeted therapies such as gefitinib are designed to inhibit oncogenic signaling pathways in EGFR-mutant non-small cell lung cancer (NSCLC). While clinical outcomes are often measured through tumor response or progression-free survival, the biological mechanisms that generate these outcomes occur across multiple interconnected scales.

Drug molecules must first diffuse through tumor tissue, creating spatially heterogeneous concentration distributions. These local concentration differences influence the degree of EGFR inhibition, which in turn affects downstream ERK signaling, cell-cycle regulation, and ultimately cellular proliferation. As a result, regions of the same tumor may experience substantially different therapeutic effects despite receiving the same treatment.

Directly measuring each stage of this transport-to-response process throughout an entire tumor is experimentally challenging. Computational modeling provides a framework for connecting these biological scales, allowing the effects of spatial drug transport to be propagated through signaling and proliferation networks.

This project develops a mechanistic systems biology model that links gefitinib diffusion, EGFR inhibition, ERK pathway activity, and proliferation in L858R-mutant NSCLC. Through a series of progressively complex experiments, the model investigates how spatial heterogeneity in drug exposure generates heterogeneous biological responses and identifies the parameters that most strongly influence therapeutic efficacy.

This leads to the central research question:

**To what extent does spatial heterogeneity in gefitinib concentration generate heterogeneous proliferation responses in L858R-mutant NSCLC through differential EGFR inhibition?**

## Scientific Background

### EGFR Signaling in L858R-Mutant NSCLC

The Epidermal Growth Factor Receptor (EGFR) regulates cellular growth, survival, and proliferation through signaling pathways such as the RAS–RAF–MEK–ERK cascade. In many cases of non-small cell lung cancer (NSCLC), activating mutations in EGFR cause persistent pathway activation and uncontrolled cell division.

One of the most common activating mutations is L858R, which increases EGFR kinase activity and makes tumor cells highly dependent on EGFR signaling. Because of this dependence, EGFR has become an important therapeutic target in NSCLC.

### Gefitinib and EGFR Inhibition

Gefitinib is a small-molecule tyrosine kinase inhibitor (TKI) that binds to the intracellular kinase domain of EGFR and suppresses its activity. Reduced EGFR activity decreases downstream ERK signaling, leading to lower Cyclin D/CDK4/6 activity and reduced cellular proliferation.

The relationship between drug concentration and pathway inhibition is nonlinear and is commonly represented using a Hill equation:

EGFR Activity = $$\frac{1}{1 + \left(\frac{C}{IC_{50}}\right)^n}$$

Where:

- `C` is local gefitinib concentration
- `IC50` is the concentration producing 50% inhibition
- `n` is the Hill coefficient

This relationship forms the pharmacodynamic component of the model.

### Fick's Second Law of Diffusion

Before a drug can inhibit its target, it must first reach the target location. Drug transport through tissue can be described using Fick's second law of diffusion:

$$
\frac{\partial C}{\partial t}
=
D \frac{\partial^2 C}{\partial x^2}
$$

Where:

- `C` is drug concentration
- `t` is time
- `x` is spatial position
- `D` is the effective diffusion coefficient

Diffusion creates concentration gradients throughout tissue, resulting in regions that experience different levels of drug exposure.

### Spatial Heterogeneity in Drug Delivery

Tumor tissue is structurally heterogeneous. Variations in extracellular matrix density, cellular packing, vascular access, and tissue architecture can alter local drug transport properties.

These differences may produce spatially heterogeneous gefitinib concentrations even within the same tumor. As a result, neighboring regions can experience substantially different levels of EGFR inhibition and downstream signaling suppression.

This project models tissue heterogeneity through spatially varying diffusion coefficients to investigate how transport differences influence therapeutic response.

### From Drug Transport to Proliferation Response

The central hypothesis of this work is that spatial heterogeneity in gefitinib concentration propagates through multiple biological scales:

Drug Transport
↓
Local Gefitinib Concentration
↓
EGFR Inhibition
↓
ERK Signaling Activity
↓
Cell-Cycle Regulation
↓
Proliferation Response

By coupling diffusion dynamics with signaling and proliferation models, the framework can quantify how physical transport processes contribute to heterogeneous therapeutic outcomes in L858R-mutant NSCLC.

## A Note on Scientific Transparency

This project uses a reduced-order mechanistic framework to connect drug transport, signaling activity, and cellular response. Several biological processes are intentionally simplified to make the system computationally tractable while preserving the primary relationships relevant to the research question.

First, EGFR inhibition is represented using a Hill equation rather than an explicit receptor-binding model. This approximation captures the experimentally observed relationship between gefitinib concentration and receptor activity but does not model individual binding and unbinding events.

Second, downstream signaling is represented as a simplified EGFR → ERK relationship rather than a full RAS–RAF–MEK–ERK pathway model. While this reduces biological detail, it preserves the key objective of tracking how changes in drug concentration propagate through the signaling network.

Third, proliferation is modeled as a function of signaling activity rather than through explicit simulation of cell-cycle dynamics, gene regulation, or population growth. The resulting proliferation values should therefore be interpreted as relative biological response rather than direct predictions of tumor size.

Finally, tissue heterogeneity is represented through spatial variation in effective diffusion coefficients. Real tumors exhibit substantially greater complexity, including heterogeneous vascularization, extracellular matrix composition, cellular density, and microenvironmental structure. The model isolates diffusion-driven heterogeneity in order to study its influence on signaling and therapeutic response.

These assumptions are documented explicitly to distinguish model predictions from biological reality and to identify opportunities for future refinement.
